```python

import random
import statistics
from typing import Dict, Any, List

def researcher_agent(topic: str) -> Dict[str, Any]:
    """Simulated literature review output."""
    summary = (
        f"Based on synthesized research, '{topic}' has been investigated across "
        "environmental and socio-economic studies. Key themes include drivers, "
        "local adaptation, and long-term trends."
    )
    key_papers = [
        "Paper 1: Climate Impact Studies (2021) – Journal of Coastal Research",
        "Paper 2: Regional Erosion Patterns (2020) – Marine GeoScience",
        "Paper 3: Human Adaptation Strategies (2022) – Environmental Policy Review",
    ]
    research_gaps = [
        "Insufficient localized longitudinal datasets",
        "Few causal studies linking emissions directly to local erosion",
        "Limited studies incorporating socio-economic adaptation variables",
    ]
    return {
        "topic": topic,
        "summary": summary,
        "key_papers": key_papers,
        "research_gaps": research_gaps,
    }

def hypothesis_generator_agent(summary: str) -> Dict[str, Any]:
    """Generate plausible hypotheses from a literature summary."""
    hypotheses = [
        "H1: Increased CO2 emissions accelerate coastal erosion in Southeast Asia.",
        "H2: Presence of coastal vegetation reduces erosion rate significantly.",
        "H3: Local mitigation policies reduce long-term erosion trends."
    ]
    # simple ranking heuristic (dummy)
    selected_hypothesis = hypotheses[0]
    rationale = "Selected because it's directly referenced by multiple reviewed papers and is testable with time-series erosion data."
    return {
        "hypotheses": hypotheses,
        "selected_hypothesis": selected_hypothesis,
        "rationale": rationale
    }

def data_analyst_agent(hypothesis: str, seed: int = 42) -> Dict[str, Any]:
    """
    Produce dummy time-series data and simple statistical analysis.
    Returns a dict with 'visualization_data' (list of dicts), 'analysis_summary', and stats.
    """
    random.seed(seed)
    years = list(range(2010, 2024))
    values = [round(random.uniform(0.5 + 0.05*(i), 2.5 + 0.05*(i)), 2) for i in range(len(years))]
    # introduce a gentle upward trend
    visualization_data = [{"Year": y, "Erosion Rate": v} for y, v in zip(years, values)]
    mean_val = round(statistics.mean(values), 3)
    stdev_val = round(statistics.pstdev(values), 3)
    analysis_summary = (
        "The dummy dataset shows a gentle upward trend in erosion rates from 2010 to 2023. "
        "This simulated trend is consistent with the selected hypothesis."
    )
    return {
        "visualization_data": visualization_data,
        "analysis_summary": analysis_summary,
        "stats": {"mean": mean_val, "std": stdev_val}
    }

def writer_editor_agent(topic: str, summary: str, hypothesis: str, analysis: Dict[str, Any]) -> str:
    """Return a markdown-style final draft string summarizing the paper."""
    md = f"""# Research Paper on {topic}

## Abstract
{summary}

## Hypothesis
{hypothesis}

## Data Analysis
{analysis['analysis_summary']}

### Statistics
- Mean erosion rate: {analysis['stats']['mean']}
- Std deviation: {analysis['stats']['std']}

## Conclusion
The simulated analysis supports the selected hypothesis with a modest upward trend in erosion rate.
"""
    return md

```

```python
%%writefile app.py
# -------------------------
# app.py - Self-contained Streamlit Dashboard
# -------------------------
import streamlit as st
import matplotlib.pyplot as plt
import pandas as pd
import random, statistics

# ---------- Module 2 Dummy Agents ----------
def researcher_agent(topic: str):
    summary = (
        f"Based on synthesized research, '{topic}' has been investigated across "
        "environmental and socio-economic studies. Key themes include drivers, "
        "local adaptation, and long-term trends."
    )
    key_papers = [
        "Paper 1: Climate Impact Studies (2021) – Journal of Coastal Research",
        "Paper 2: Regional Erosion Patterns (2020) – Marine GeoScience",
        "Paper 3: Human Adaptation Strategies (2022) – Environmental Policy Review",
    ]
    research_gaps = [
        "Insufficient localized longitudinal datasets",
        "Few causal studies linking emissions directly to local erosion",
        "Limited studies incorporating socio-economic adaptation variables",
    ]
    return {"topic": topic, "summary": summary, "key_papers": key_papers, "research_gaps": research_gaps}

def hypothesis_generator_agent(summary: str):
    hypotheses = [
        "H1: Increased CO2 emissions accelerate coastal erosion in Southeast Asia.",
        "H2: Presence of coastal vegetation reduces erosion rate significantly.",
        "H3: Local mitigation policies reduce long-term erosion trends."
    ]
    selected_hypothesis = hypotheses[0]
    rationale = "Selected because it's directly referenced by multiple reviewed papers and is testable with time-series erosion data."
    return {"hypotheses": hypotheses, "selected_hypothesis": selected_hypothesis, "rationale": rationale}

def data_analyst_agent(hypothesis: str, seed: int = 42):
    random.seed(seed)
    years = list(range(2010, 2024))
    values = [round(random.uniform(0.5 + 0.05*(i), 2.5 + 0.05*(i)), 2) for i in range(len(years))]
    visualization_data = [{"Year": y, "Erosion Rate": v} for y, v in zip(years, values)]
    mean_val = round(statistics.mean(values), 3)
    stdev_val = round(statistics.pstdev(values), 3)
    analysis_summary = (
        "The dummy dataset shows a gentle upward trend in erosion rates from 2010 to 2023. "
        "This simulated trend is consistent with the selected hypothesis."
    )
    return {"visualization_data": visualization_data, "analysis_summary": analysis_summary, "stats": {"mean": mean_val, "std": stdev_val}}

def writer_editor_agent(topic: str, summary: str, hypothesis: str, analysis: dict):
    md = f"""# Research Paper on {topic}

## Abstract
{summary}

## Hypothesis
{hypothesis}

## Data Analysis
{analysis['analysis_summary']}

### Statistics
- Mean erosion rate: {analysis['stats']['mean']}
- Std deviation: {analysis['stats']['std']}

## Conclusion
The simulated analysis supports the selected hypothesis with a modest upward trend in erosion rate.
"""
    return md

# ---------- STREAMLIT CONFIG ----------
st.set_page_config(page_title="Dynamic Academic Research Dashboard", layout="wide")

# ---------- CUSTOM STYLES ----------
st.markdown("""
<style>
body { background-color: #0f172a; }
.main-title { text-align:center; font-size:2.5rem; font-weight:800; color:white; margin-top:40px; }
.topnav { display:flex; justify-content:flex-end; align-items:center; background-color:#0f172a; color:white;
          padding:1rem 2rem; font-size:1rem; gap:1.5rem; border-bottom:1px solid #1e293b; }
.topnav a { color:#cbd5e1; text-decoration:none; transition:0.3s; }
.topnav a:hover { color:white; text-decoration:underline; }
.info-box { background-color:#1e293b; border-radius:10px; padding:20px; color:#cbd5e1; text-align:center; margin-top:40px; }
</style>
""", unsafe_allow_html=True)

# ---------- TOP NAVIGATION ----------
st.markdown("""
<div class="topnav">
    <a href="?page=home">🏠 Dashboard</a>
    <a href="?page=workspace">📝 Writing Workspace</a>
    <a href="?page=signin">🔐 Sign In</a>
</div>
""", unsafe_allow_html=True)

page = st.query_params.get("page", ["home"])[0]

# ---------- HOME PAGE ----------
if page == "home":
    st.markdown("<div class='main-title'>Dynamic Academic Research Dashboard</div>", unsafe_allow_html=True)
    topic = st.text_input("Enter your research topic...", placeholder="e.g., Impact of AI on Education")
    if st.button("Start Research"):
        st.session_state["topic"] = topic or "AI in Education"
        st.success(f"Topic selected: *{st.session_state['topic']}*")

    st.divider()
    st.subheader("Quick Actions")
    col1, col2, col3 = st.columns(3)
    with col1:
        if st.button("📚 Literature Review"):
            st.query_params["page"] = ["literature"]
    with col2:
        if st.button("💡 Hypothesis"):
            st.experimental_set_query_params(page="hypothesis")
    with col3:
        if st.button("📊 Analysis"):
            st.experimental_set_query_params(page="analysis")

    st.markdown("<div class='info-box'>👉 Enter a topic and choose a section to begin simulated research.</div>", unsafe_allow_html=True)

# ---------- LITERATURE REVIEW ----------
elif page == "literature":
    topic = st.session_state.get("topic", "AI in Education")
    st.subheader(f"📚 Literature Review for: {topic}")
    research_output = researcher_agent(topic)
    st.markdown(research_output["summary"])
    st.markdown("**Key papers:**\n- " + "\n- ".join(research_output["key_papers"]))
    st.markdown("**Research gaps:**\n- " + "\n- ".join(research_output["research_gaps"]))

# ---------- HYPOTHESES ----------
elif page == "hypothesis":
    topic = st.session_state.get("topic", "AI in Education")
    st.subheader(f"💡 Hypotheses for: {topic}")
    hypothesis_output = hypothesis_generator_agent("dummy summary")
    for h in hypothesis_output["hypotheses"]:
        st.markdown(f"- {h}")
    st.markdown(f"**Selected hypothesis:** {hypothesis_output['selected_hypothesis']}")
    st.markdown(f"*Rationale:* {hypothesis_output['rationale']}")

# ---------- ANALYSIS ----------
elif page == "analysis":
    topic = st.session_state.get("topic", "AI in Education")
    st.subheader(f"📊 Analysis for: {topic}")
    analysis = data_analyst_agent("H1: Dummy hypothesis")
    st.markdown("### Analysis Summary")
    st.markdown(analysis["analysis_summary"])

    # Plot dummy data
    df = pd.DataFrame(analysis["visualization_data"])
    st.line_chart(df.rename(columns={"Year": "index"}).set_index("index"))

    st.markdown(f"- Mean: {analysis['stats']['mean']}")
    st.markdown(f"- Std dev: {analysis['stats']['std']}")

# ---------- WORKSPACE ----------
elif page == "workspace":
    st.title("📝 Writing Workspace")
    topic = st.session_state.get("topic", "AI in Education")
    final_draft = writer_editor_agent(topic, "dummy summary", "H1: Dummy hypothesis", data_analyst_agent("H1: Dummy hypothesis"))
    st.text_area("Draft Paper", value=final_draft, height=300)

# ---------- SIGN IN ----------
elif page == "signin":
    st.title("🔐 Sign In")
    username = st.text_input("Username")
    password = st.text_input("Password", type="password")
    if st.button("Login"):
        st.success("✅ Logged in successfully!")
```
